## 13. Лінійна алгебра

### Матричний добуток

```python
A = np.array([
    [1, 2, 3],
    [0, 1, 2]
])

B = np.array([
    [ 1, 2],
    [ 0, 4],
    [-1, 0]
])

C = A @ B
print(C)
```

### Діагональ, транспонування та визначник

```python
A = np.array([
    [ 1.,  0., 3.],
    [-1., -1., 2.],
    [ 4.,  7., 2.]
])

print("diag:", np.diag(A))
print("\nA.T =\n", A.T)

det = np.linalg.det(A)
print("\ndet(A) =", det)
print("det близький до -25:", np.isclose(det, -25.0))
```

### Розв’язання системи `Ax = b`

```python
A = np.array([
    [1.,  2.,  5.],
    [1., -1.,  3.],
    [3., -6., -1.]
])

b = np.array([-9., 2., 25.])

x = np.linalg.solve(A, b)

print("x =", x)
print("Перевірка A @ x ≈ b:", np.allclose(A @ x, b))
```

Для розв’язання системи не потрібно явно обчислювати `A⁻¹`.  
У числових обчисленнях перевага надається `np.linalg.solve(A, b)`.


### Метод найменших квадратів

```python
X = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.],
    [1., 4.]
])

y = np.array([2.1, 3.9, 6.2, 8.1])

beta, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)

print("beta:", beta)
print("residuals:", residuals)
print("rank:", rank)
print("singular values:", s)
```

### Власні значення та власні вектори

```python
A = np.array([
    [2., 1.],
    [1., 2.]
])

values, vectors = np.linalg.eigh(A)

print("Власні значення:", values)
print("Власні вектори:\n", vectors)
```

### SVD

```python
A = np.array([
    [3., 1.],
    [1., 3.],
    [1., 1.]
])

U, s, Vt = np.linalg.svd(A, full_matrices=False)

print("U =\n", U)
print("\ns =", s)
print("\nVt =\n", Vt)
```

## 18. Узагальнюючий приклад: підготовка даних і лінійний прогноз

```python
# 1. Вхідні дані: два фактори та цільова змінна
X = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(X).all(axis=1) & np.isfinite(y)
X = X[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = X.mean(axis=0)
sigma = X.std(axis=0)
X_std = (X - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
X_design = np.column_stack([
    np.ones(X_std.shape[0]),
    X_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(X_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = X_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

## 19. Підсумкові тренувальні завдання

1. Створіть матрицю `X` форми `(5, 4)` із випадкових чисел.
2. Обчисліть середнє та стандартне відхилення кожної ознаки.
3. Стандартизуйте матрицю через broadcasting.
4. Виберіть рядки, у яких значення першої ознаки додатне.
5. Додайте до матриці новий стовпець одиниць.
6. Створіть вектор ваг і обчисліть прогноз `X @ w`.
7. Для заданого `y_true` обчисліть MSE.
8. Перевірте, чи немає в масивах `NaN` або `inf`.


```python
# Виконайте підсумкове завдання тут
rng = np.random.default_rng(seed=42)

# 1. Створення випадкової матриці X (5, 4) та цільового вектора y_true
X = rng.standard_normal((5, 4))
y_true = np.array([1.5, 2.0, -0.5, 3.2, 1.1])

print("Матриця X:\n", X)
print("\nВектор y_true:", y_true)

# 2. Перевірка на відсутність NaN або inf у масивах
d= not np.all(np.isfinite(X)) or not np.all(np.isfinite(y_true))
print("\nПеревірка на NaN & inf:", d)

# Фільтрація валідних рядків
mask = np.isfinite(X).all(axis=1) & np.isfinite(y_true)
X = X[mask]
y_true = y_true[mask]

# 3. Вибір рядків, у яких значення першої ознаки додатне
p_mask = X[:, 0] > 0
X_fil = X[p_mask]
y_fil = y_true[p_mask]

print("\nБулева маска (перша ознака > 0):", p_mask)
print("\nВідфільтрована матриця X:\n", X_fil)
print("\nВідфільтровані значення y_true:", y_fil)

# 4. Обчислення середнього та стандартного відхилення кожної ознаки
mu = X_fil.mean(axis=0)
sigma = X_fil.std(axis=0)

print("\nСереднє значення (mu):", mu)
print("Стандартне відхилення (sigma):", sigma)

# 5. Стандартизація матриці через broadcasting
X_std = (X_fil - mu) / sigma

print("\nСтандартизована матриця X_std:\n", X_std)

# 6. Додавання стовпця одиниць
random_column = rng.standard_normal(X_std.shape[0])
X_design = np.column_stack([random_column, X_std])

print("\nМатриця з доданим стовпцем одиниць:\n", X_design)

# 7. Створення вектора ваг w та обчислення прогнозу
w = np.array([0.5, -1.2, 0.8, 0.3, -0.5])
y_pred = X_design @ w

print("\nВектор ваг w:", w)
print("Прогнозовані значення (y_pred):", y_pred)

# 8. Обчислення MSE
mse = np.mean((y_fil - y_pred) ** 2)

print("\nMSE:", mse)
```
